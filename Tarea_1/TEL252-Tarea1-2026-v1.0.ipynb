* **Algoritmo de Euclides:** algoritmo histórico para encontrar un valor máximo común divisor (MCD) de dos números enteros. 
* **Algoritmo de Euclides Extendido:** Permite una búsqueda eficiente de factores relativamente primos. También, es muy relevante en criptografía de llave pública.

```python
def gcd(a, b):
    while a != 0:
        a, b = b % a, a
    return b

def eea(r0,r1):
    if r0 == 0:
        return (r1, 0, 1)
    else:
        g, s, t = eea(r1 % r0, r0)
        return (g, t - (r1 // r0) * s, s)

r0, r1 = 13, 9
print(gcd(r0,r1))
print(eea(r0,r1))
```

## Lógica ROT47

Este código de cifrado funciona de misma forma que el ROT13, rotando la letra en clave de 47 en vez de 13

### Diferencia de ROT13

A diferencia de ROT13, además de la clave, ROT47 funciona en las clave ASCCI, con los caracteres desde 33 a 126. (Desde el 1 al 32 son invisibles, al igual que el 127). Esto nos obliga a usar el mod 94 para estar dentro de ese rango de caracteres

## Código de ejemplo: 

```python
def rot(texto, n ):
    resultado = []
    for x in texto:
        if 33 <= ord(x) <= 126:
            resultado.append(chr(33 + ((ord(x) -33 + n) % 94)))
        else:
            resultado.append(x)
    return ''.join(resultado)

def rot47(texto):
    return rot(texto, 47)
```

```python
# Escriba aquí su código y markdown
#La debilidad del cifrado de sustitucion Affine es que el espacio de claves es muy pequeño, teniendo solo 6840 claves disponibles
import string
import math

alpha = string.ascii_letters+string.digits+string.punctuation+' '#Tenemos 95 posibles caracteres
m = len(alpha) 


def decrypt(a, b, ciphertext):
    # Inverso multiplicativo de a modulo m
    a_inv = pow(a, -1, m)

    plaintext = ''

    for symbol in ciphertext:
        if symbol in alpha:
            c = alpha.find(symbol)

            # x = a^(-1)(c-b) mod m
            x = (a_inv * (c - b)) % m

            plaintext += alpha[x]
        else:
            plaintext += symbol

    return plaintext


def brute_force(ciphertext):
    candidates = []

    for a in range(m):
        # Las claves para affine requieren cumplir que  gcd(a,m)=1
        if math.gcd(a, m) != 1:
            continue

        for b in range(m):
            plaintext = decrypt(a, b, ciphertext)

            candidates.append((a, b, plaintext))

    return candidates


import datetime

def main():
    ciphertext = input("Ingrese el texto cifrado: ")
    keyword = input("Ingrese una palabra para filtrar (Enter para guardar todos en un txt): ").strip().upper()

    candidates = brute_force(ciphertext)

    if keyword:
        matches = [c for c in candidates if keyword in c[2].upper()]
        
        print("\n" + "="*50)
        print(f"Claves totales probadas: {len(candidates)}")
        print(f"Coincidencias mostradas: {len(matches)}")
        print("="*50 + "\n")

        for a, b, plaintext in matches:
            print(f"[a={a:2d}, b={b:2d}] -> {plaintext}")
            
        if not matches:
            print("No se encontraron resultados con esa palabra clave.")
            
    else:
        fecha_actual = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
        nombre_archivo = f"affine_{fecha_actual}.txt"
        
        try:
            with open(nombre_archivo, "w", encoding="utf-8") as file:
                file.write(f"Texto cifrado original: {ciphertext}\n")
                file.write(f"Claves probadas: {len(candidates)}\n")
                file.write("="*50 + "\n")
                
                for a, b, plaintext in candidates:
                    file.write(f"[a={a:2d}, b={b:2d}] -> {plaintext}\n")
                    
            print(f"\n[+] Éxito: Se han guardado {len(candidates)} combinaciones en '{nombre_archivo}'")
            
        except IOError as e:
            print(f"\n[-] Error al intentar guardar el archivo: {e}")

if __name__ == '__main__':
    main()
```

### **Actividad 4**: Construir un registro de desplazamiento de retroalimentación lineal eficiente para un campo finito binario, que cumpla con:
- Utilizar un polinomio primitivo de grado mayor a 20.
- Demuestre que la llave de flujo es consistente con el periodo máximo.
- No es posible invocar funciones LFSR preexistentes p.e. PyLFSR.

```python
import secrets
```

```python
grado = 21
max_val = 2**(grado-1)

llave_inicial = secrets.randbelow(max_val) + 1 # Para evitar el 0
polinomio = (1 << 0) | (1 << 2)  # (0,1,21)

print(f"Llave inicial: {llave_inicial}")
for i in range(grado-1,-1, -1):
    print((llave_inicial>>i)&1, end="")
print("\n")
print(f"Polinomio: {polinomio}")
for i in range(grado-1,-1,-1):
    print((polinomio>>i)&1, end="")

#Cheat sheet de operaciones bitwise:
#>> shift derecho
#<< shift izquierdo
#& AND
#^ XOR

# Inicializar variable de salida
output = llave_inicial&1
registros = llave_inicial
clock_cycle = 0

def calcular_output():
    global clock_cycle
    global polinomio
    global registros
    global output
    
    if clock_cycle == 0:
        clock_cycle+=1
        return output
    else:
        suma = output&(polinomio&1)
        for j in range(0,grado):
            polinomio_j = (polinomio>>j)&1
            registro_j = (registros>>j)&1
            suma ^= polinomio_j&registro_j
        registros >>= 1
        registros = registros | (suma << (grado-1))
        clock_cycle+=1
        return(registros&1)

def imprimir_registros_polinomio():
    global grado
    global registros
    global polinomio

    print(f"Registros: ", end="")
    for i in range(grado-1,-1,-1):
        print((registros>>i)&1, end="")
    print(", ",end="")
    print(f"Polinomio: ", end="")
    for i in range(grado-1,-1,-1):
        print((polinomio>>i)&1, end="")
    print("")
```

```python
# --- INICIO DEL CHECK DE LONGITUD MÁXIMA ---
#Nota: La implementación de las funciones del bloque anterior fueron hechas 100% por humanos.
# el único aporte de la IA fueron para el código a continuación, así como para consolidar y aprender los conceptos necesarios, como el manejo
# de las operaciones bitwise, y comprensión del texto Understanding Cryptography - Cristof Paar, Jan Pelzl.

print("-" * 30)
print("Iniciando check de periodo...")

periodo_teorico = (2 ** grado) - 1
ciclos_reales = 0

# Guardamos el objetivo
estado_objetivo = llave_inicial

while True:
    calcular_output()
    #imprimir_registros_polinomio()
    
    # Tu función calcular_output() no altera 'registros' en clock_cycle == 0,
    # por lo que empezamos a contar y verificar a partir del ciclo 1.
    if clock_cycle > 1:
        ciclos_reales += 1
        
        # Si el registro vuelve a la llave inicial, completamos el ciclo
        if registros == estado_objetivo:
            break
            
        # Freno de emergencia: Si supera el periodo teórico, 
        # significa que entró en un sub-ciclo y nunca volverá a la llave inicial.
        if ciclos_reales > periodo_teorico:
            break

print(f"Periodo alcanzado : {ciclos_reales}")
print(f"Periodo máximo teórico : {periodo_teorico}")

if ciclos_reales == periodo_teorico:
    print("✅ ¡ÉXITO! El LFSR alcanzó la longitud máxima.")
    print("   (El polinomio generado ES PRIMITIVO).")
else:
    print("❌ NO alcanzó la longitud máxima.")
    print("   (El polinomio generado NO es primitivo o el estado se atascó).")
```
