se realizó un análisis estadístico del texto cifrado para identificar la letra con mayor ocurrencia. Este enfoque se adoptó porque al inicio de la actividad se nos proporcionó la distribución de frecuencias de los caracteres en inglés, lo que nos permitió ejecutar un ataque por análisis de frecuencia, nos dio como resultado que la letra más repetida en el texto cifrado fue la 'w', esto nos da un primer acercamiento para conocer el texto real 

```python
import pandas as pd

# 2. Lectura del archivo
parafo = []
linea = []
count = 0
with open("textocifrado.txt") as file:
    for line in file:
        linea.append(count)
        parafo.append(line.strip())
        count += 1
data = {
    "linea": linea[:len(linea)-1],
    "parafo": parafo[:len(parafo)-1]
}
df = pd.DataFrame(data)


frecuencias = []
for letra in en['let']:
    frec = df['parafo'].str.count(letra).sum()
    frecuencias.append(frec)

dict_frec = {'letra': en['let'], 'frec': frecuencias}
df_frecuencias = pd.DataFrame(dict_frec)

df_frecuencias_ordenado = df_frecuencias.sort_values(by='frec', ascending=False).reset_index(drop=True)

diccionario_mapeo = {}
for index, row in df_frecuencias_ordenado.iterrows():
    if index < len(en['let']):
        letra_cifrada = row['letra']
        letra_ingles = en['let'][index] 
        diccionario_mapeo[letra_cifrada] = letra_ingles

def descifrar_linea(texto):
    texto_claro = ""
    for caracter in texto:
        if caracter in diccionario_mapeo:
            texto_claro += diccionario_mapeo[caracter]
        else:
            texto_claro += caracter 
    return texto_claro


for linea in df['parafo']:
    print(descifrar_linea(linea))
```


en esta primera iteración se puede observar la repetición de la palabra 'tre', sabiendo que en el idioma inglés la palabra de tres letras más frecuente es 'the', podemos deducir que las asignaciones de 't' y 'e' son correctas, pero debemos corregir manualmente que 'r' = 'h'. Además, al analizar la primera palabra del texto cifrado ('yoc'),nos indica que se trata de  'you'; esto nos confirma que 'y' = 'y', 'o' = 'o', y nos revela una nueva letra: 'c' = 'u'. Al aplicar esta misma lógica deductiva al resto del texto, logramos construir el siguiente diccionario corregido:


```python
correcciones_manuales = {
    'c': 'a', 'i': 'b', 'p': 'c', 'g': 'd', 'w': 'e', 
    'a': 'f', 'h': 'g', 'e': 'h', 'k': 'i', 'b': 'j', 
    'n': 'k', 'v': 'l', 's': 'm', 't': 'n', 'm': 'o', 
    'f': 'p', 'd': 'q', 'x': 'r', 'j': 's', 'y': 't', 
    'q': 'u', 'o': 'v', 'u': 'w', 'r': 'x', 'l': 'y', 
    'z': 'z'
}
diccionario_mapeo.update(correcciones_manuales)
for linea in df['parafo']:
    print(descifrar_linea(linea))
```

* **Algoritmo de Euclides:** algoritmo histórico para encontrar un valor máximo común divisor (MCD) de dos números enteros. 
* **Algoritmo de Euclides Extendido:** Permite una búsqueda eficiente de factores relativamente primos. También, es muy relevante en criptografía de llave pública.

```python
def gcd(a, b):
    while a != 0:
        a, b = b % a, a
    return b

def eea(r0,r1):
    if r0 == 0:
        return (r1, 0, 1)
    else:
        g, s, t = eea(r1 % r0, r0)
        return (g, t - (r1 // r0) * s, s)

r0, r1 = 13, 9
print(gcd(r0,r1))
print(eea(r0,r1))
```

* **XOR o Exclusive OR:** es una operación lógica que denota una desigualdad, es decir, el resultado es verdadero si los valores son distintos, de otra forma es falso. También, representa una adición módulo 2. En Python el operador `^` ("_caret_") es equivalente al operador XOR, mientras que el símbolo `&` representa al operador lógico AND. 

```python
import numpy as np
a = np.array([[0,0],[1,1]])
x = np.array([[1,0],[0,1]])
print(a ^ x) # bitwise XOR
print(a & x) # bitwise AND
```

* **Función Phi de Euler o Totient de Euler:** Identifica cuántos números coprimos existen en un campo finito dado. La función Phi(n) es iterativa utilizando una factorización canónica de números primos.

```python
from math import gcd
def myeuler(m):
    phi = 0
    for a in range(1,m):
        if (gcd(a,m) == 1): 
            phi += 1
    return phi
Zm = 81
print("Phi(m): ",myeuler(Zm))
```
